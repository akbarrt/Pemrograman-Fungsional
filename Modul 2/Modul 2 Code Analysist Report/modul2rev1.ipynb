## Analysis Question 1:
Mengapa add_bonus2() mengubah data asli exam_scores meskipun tidak
menggunakan kata kunci global? Jelaskan hasil analisis kamu. Berdasarkan
karakteristiknya, apakah fungsi add_bonus2() termasuk fungsi pure atau impure?
Jelaskan alasannya?

```python
# data awal (sama seperti Gambar 2.3 pada modul)
exam_scores = [70, 80, 85]
print("Original exam_scores:", exam_scores)
```

```python
def add_bonus2(bonus):
    for i in range(len(exam_scores)):
        exam_scores[i] = exam_scores[i] + bonus
    return exam_scores
```

**Run ke-1** (bonus = 5)

```python
# 202410370110388
print("Before:", exam_scores)
result = add_bonus2(5)
print("New result:", result)
print("Original exam_scores be:", exam_scores)
```

**Run ke-2** (cell yang sama dijalankan lagi, tanpa mereset `exam_scores`)

```python
# 202410370110388
print("Before:", exam_scores)
result = add_bonus2(5)
print("New result:", result)
print("Original exam_scores be:", exam_scores)
```

**Run ke-3**

```python
# 202410370110388
print("Before:", exam_scores)
result = add_bonus2(5)
print("New result:", result)
print("Original exam_scores be:", exam_scores)
```

## Jawaban:
**Pengamatan.** Pada setiap run, input fungsi sama (`bonus = 5`), tetapi hasilnya berbeda: `[75, 85, 90]` → `[80, 90, 95]` → `[85, 95, 100]`. Nilai `exam_scores` yang asli ikut berubah dan bonus menumpuk setiap kali fungsi dipanggil.

**Mengapa data asli berubah tanpa kata kunci `global`?**
1. Kata kunci `global` hanya diperlukan jika sebuah fungsi ingin **mengganti (rebind) nama variabel** global ke objek lain (misalnya `exam_scores = [...]`). Untuk **membaca** variabel global, Python otomatis mencarinya di scope luar (aturan LEGB), sehingga `exam_scores` bisa diakses dari dalam fungsi walaupun tidak dideklarasikan `global`.
2. Pada `exam_scores[i] = exam_scores[i] + bonus`, fungsi tidak mengganti nama `exam_scores`. Fungsi hanya **memodifikasi isi** objek list yang ditunjuk oleh nama tersebut (in-place mutation). List adalah objek **mutable** dan hanya ada satu objek list di memori, sehingga perubahan dari dalam fungsi langsung terlihat di luar fungsi.
3. Jadi `global` bukan satu-satunya jalan untuk menimbulkan side effect. Mutasi objek mutable juga bisa melakukannya, dan efeknya lebih tersembunyi karena tidak ada kata kunci yang memberi tanda.

**Apakah pure atau impure? → Impure Function**, karena:
- **Tidak deterministik**: input yang sama (`5`) menghasilkan output berbeda pada setiap pemanggilan, karena hasilnya bergantung pada kondisi `exam_scores` saat itu.
- **Bergantung pada state eksternal**: fungsi tidak menerima data lewat parameter, melainkan membaca variabel `exam_scores` di luar fungsi.
- **Memiliki side effect**: fungsi mengubah data di luar scope-nya (`exam_scores` asli berubah).

Agar menjadi pure function, data harus dikirim lewat parameter dan fungsi harus membuat list baru, seperti `add_bonus(scores, bonus)` pada modul, atau memakai `copy` sebelum memodifikasi.

## Exercise 2: Data duplication with shallow and deep copy

# Instruksi:
1. Buat duplikat dari variabel user menggunakan Shallow Copy (simpan
dalam variabel bernama user_shallow) dan Deep Copy (simpan dalam
variabel bernama user_deep).
2. Pada masing-masing objek hasil salinan (user_shallow dan user_deep),

3. Periksa alamat memori dari indeks yang telah dimodifikasi, lalu

## Analysis Question 2
Amati output data asli beserta alamat memorinya, kemudian jelaskan mengapa
hasil Shallow Copy dan Deep Copy berbeda!

```python
import copy

user = { "name": "putri", "scores" : [90, 85]}
```

```python
# 202410370110388
# 1. Buat duplikat
user_shallow = copy.copy(user)
user_deep = copy.deepcopy(user)

# Kondisi awal sebelum modifikasi
print("Data awal  :", user)
print("id(user['scores'])         :", id(user["scores"]))
print("id(user_shallow['scores']) :", id(user_shallow["scores"]))
print("id(user_deep['scores'])    :", id(user_deep["scores"]))
print("id(user['scores'][0])      :", id(user["scores"][0]), "(nilai 90 asli)")
```

```python
# 202410370110388
# 2. Ubah scores 90 -> 75 pada Deep Copy dulu
user_deep["scores"][0] = 75

print("=== Setelah Deep Copy dimodifikasi ===")
print("user      :", user)
print("user_deep :", user_deep)
print("id(user['scores'][0])      :", id(user["scores"][0]))
print("id(user_deep['scores'][0]) :", id(user_deep["scores"][0]))
print("Alamat sama?               :", id(user["scores"][0]) == id(user_deep["scores"][0]))
```

```python
# 202410370110388
# 2. Ubah scores 90 -> 75 pada Shallow Copy
user_shallow["scores"][0] = 75

print("=== Setelah Shallow Copy dimodifikasi ===")
print("user         :", user)
print("user_shallow :", user_shallow)
print("id(user['scores'][0])         :", id(user["scores"][0]))
print("id(user_shallow['scores'][0]) :", id(user_shallow["scores"][0]))
print("Alamat sama?                  :", id(user["scores"][0]) == id(user_shallow["scores"][0]))
print("List scores sama objek?       :", user["scores"] is user_shallow["scores"])
```

## Jawaban:
**Hasil pengamatan**
- **Deep Copy**: setelah `user_deep["scores"][0] = 75`, data `user` asli **tetap** `[90, 85]`. Alamat memori `scores` pada `user_deep` berbeda dengan `user`, dan alamat elemen index `[0]` juga berbeda.
- **Shallow Copy**: setelah `user_shallow["scores"][0] = 75`, data `user` asli **ikut berubah** menjadi `[75, 85]`. Alamat memori `scores` dan elemen index `[0]` pada `user_shallow` sama dengan `user`.

**Penjelasan**
- `copy.copy()` (shallow) hanya membuat **dictionary luar yang baru**. Isi dictionary, yaitu key `"scores"`, hanya disalin **referensinya**. Karena itu `user["scores"]` dan `user_shallow["scores"]` menunjuk ke **objek list yang sama** di memori. Mengubah isi list lewat salinan juga mengubah data asli. Ini melanggar prinsip immutability dan Effect-Free Programming.
- `copy.deepcopy()` menyalin objek secara **rekursif** sampai ke elemen terdalam, sehingga list `scores` juga diduplikasi menjadi objek baru dengan alamat memori sendiri. Perubahan pada `user_deep` tidak memengaruhi `user`.
- Karena `user` adalah data nested (dictionary yang berisi list), Deep Copy adalah pilihan yang tepat untuk menjaga data asli. Shallow Copy hanya aman untuk data flat.

*Catatan:* angka kecil seperti 75 dan 90 adalah integer immutable yang di-cache oleh Python. Karena itu alamat memori elemen `75` bisa sama di beberapa tempat. Yang perlu dibandingkan adalah alamat list `scores`, dan apakah data `user` asli ikut berubah.

## Exercise 3: Using iter() and next() to Process Tuple Data
Buat sebuah tuple bernama sensor_data = (23.5, 24.0, 22.8, 21.1). Ubah tuple
tersebut menjadi objek iterator menggunakan fungsi iter(). Ambil dan tampilkan dua
nilai pertama dari iterator menggunakan fungsi next(), kemudian tampilkan data yang
tersisa dan belum diambil menggunakan for loop biasa!

## Analysis Question 3
1. Amati output dari for loop. Mengapa for loop langsung menampilkan data
yang tersisa (data ketiga) alih-alih melakukan iterasi dari elemen pertama?
2. Jelaskan bagaimana mekanisme for loop bekerja di balik layar dalam Python
saat menangani objek iterator tanpa menggunakan fungsi next() secara
eksplisit?

```python
# 202410370110388
sensor_data = (23.5, 24.0, 22.8, 21.1)

# Ubah tuple menjadi iterator
sensor_iterator = iter(sensor_data)
print("Tipe objek:", type(sensor_iterator))

# Ambil dua nilai pertama dengan next()
print("Data ke-1:", next(sensor_iterator))
print("Data ke-2:", next(sensor_iterator))

# Tampilkan data yang tersisa dengan for loop biasa
print("Data yang tersisa:")
for value in sensor_iterator:
    print(value)
```

## Jawaban:
**1. Mengapa for loop langsung menampilkan data ketiga?**
Iterator menyimpan **posisi (state) penelusuran** secara internal. Dua pemanggilan `next()` sudah mengambil elemen pertama (23.5) dan kedua (24.0), sehingga pointer internal berada di elemen ketiga. Iterator hanya bergerak **maju satu arah** dan tidak otomatis kembali ke awal. Ketika `for` dijalankan pada `sensor_iterator`, ia melanjutkan dari posisi terakhir, sehingga yang tampil hanya `22.8` dan `21.1`. Jika `for` dijalankan pada tuple aslinya (`sensor_data`), semua elemen akan tampil dari awal, karena setiap `iter()` pada tuple membuat iterator baru.

**2. Mekanisme for loop di balik layar**
`for x in obj:` bekerja kira-kira seperti kode berikut:
1. Python memanggil `iter(obj)` untuk mendapatkan iterator. Jika `obj` sudah berupa iterator, `iter()` mengembalikan **objek iterator yang sama** (bukan salinan baru dan tidak di-reset), sehingga posisinya tetap dilanjutkan.
2. Python memanggil `next()` pada iterator secara berulang dan memasukkan nilainya ke variabel `x`, lalu menjalankan badan loop.
3. Ketika iterator kehabisan data, `next()` memunculkan `StopIteration`. `for` menangkap exception ini secara otomatis (tanpa menampilkan error) dan loop berhenti.

Jadi `for` pada dasarnya adalah `iter()` + `next()` + penanganan `StopIteration` yang sudah dibungkus otomatis. Contoh ekuivalennya ada di cell berikut.

```python
# 202410370110388
# Ekuivalen for loop secara manual (untuk membuktikan mekanismenya)
sensor_iterator2 = iter(sensor_data)
next(sensor_iterator2)   # ambil 2 data pertama
next(sensor_iterator2)

it = iter(sensor_iterator2)          # iter() pada iterator -> objek yang sama
print("iter(iterator) adalah objek yang sama?", it is sensor_iterator2)

while True:
    try:
        value = next(it)             # yang dilakukan 'for' secara implisit
    except StopIteration:            # yang ditangkap 'for' secara implisit
        break
    print(value)
```

## Exercise 4: Convert to List Comprehension
Ubah kode berikut menjadi List Comprehension:

```python
# Kode asli (imperatif)
result = []
for letter in "universitas":
    if letter not in "aiueo":
        result.append(letter.upper())
print(result)
```

```python
# 202410370110388
# Versi List Comprehension (deklaratif)
result = [letter.upper() for letter in "universitas" if letter not in "aiueo"]
print(result)
```

Penjelasan komponen: **expression** = `letter.upper()`, **item** = `letter`, **iterable** = `"universitas"`, **condition** = `letter not in "aiueo"` (hanya huruf konsonan).

Output yang benar adalah `['N', 'V', 'R', 'S', 'T', 'S']`. Pada soal asli tertulis `'s'` (huruf kecil) di akhir, padahal `.upper()` mengubah semua huruf menjadi kapital.

## Exercise 5: Create Generator Expression
Buat sebuah Generator Expression yang menghasilkan angka dari 1 hingga 5,
lalu tampilkan nilainya menggunakan next() di dalam while loop. Tambahkan
exception handling untuk menangani kondisi ketika data sudah habis (StopIteration)
dan tampilkan pesan yang menunjukkan bahwa iterator telah selesai

```python
# 202410370110388
numbers_gen = (x for x in range(1, 6))
print("Tipe objek:", type(numbers_gen))

while True:
    try:
        value = next(numbers_gen)
        print("Nilai:", value)
    except StopIteration:
        print("Iterator telah selesai: tidak ada data yang tersisa.")
        break
```

## Exercise 6: Create Generator

Skenario: Sebuah server mencatat aktivitas sign-in menggunakan format
string tertentu. Dalam lingkungan produksi, file log dapat berkembang hingga
berukuran gigabyte, sehingga data tidak boleh dimuat sepenuhnya ke dalam RAM
menggunakan List. Kita perlu membuat fungsi “lazy” untuk memfilter hanya log
darurat.
Diberikan data raw log stream berikut (disimulasikan):

```python
server_logs = [
"INFO: User login successful",
"ERROR: Database connection failed",
"INFO: Page rendered in 25ms",
"ERROR: Unauthorized access attempt",
"INFO: User logout successful"
]
```

Instruksi:
1. Buat Generator Function bernama filter_emergency_log(log_list) yang
menerima data log.
2. Fungsi harus mengevaluasi setiap baris log. Jika sebuah baris log diawali
dengan keyword "ERROR", hasilkan (yield) baris log tersebut secara bertahap.
3. Pada program utama, panggil Generator Function tersebut dan simpan
hasilnya ke dalam variabel bernama error_bucket.
4. Ambil nilai pertama dan kedua secara manual menggunakan fungsi next().

```python
def filter_emergency_log(log_list):
    """
    pre : log_list adalah sequence/iterable berisi string log
    post: menghasilkan (yield) satu per satu baris log yang diawali "ERROR"
    """
    for log in log_list:
        if log.startswith("ERROR"):
            yield log
```

```python
# 202410370110388
# Program utama
error_bucket = filter_emergency_log(server_logs)
print("Tipe objek:", type(error_bucket))   # generator, belum ada log yang diproses

# Ambil nilai pertama dan kedua secara manual
print("Log darurat ke-1:", next(error_bucket))
print("Log darurat ke-2:", next(error_bucket))
```

## Analysis Question 4
Dari Exercise 6, jika volume data server log meningkat menjadi 10.000.000 baris,
mengapa penggunaan Generator Function dengan yield jauh lebih aman untuk memori
komputer dibandingkan dengan fungsi biasa yang mengumpulkan seluruh hasil ke
dalam list dan mengembalikannya sekaligus menggunakan return?

## Jawaban:
- **Fungsi biasa dengan `return` (eager)**: fungsi harus memproses seluruh 10.000.000 baris, mengumpulkan semua baris ERROR ke dalam sebuah list, lalu mengembalikannya sekaligus. List tersebut harus tersimpan penuh di RAM. Jika banyak baris yang cocok, pemakaian memori bisa membengkak hingga RAM habis dan program crash (*MemoryError*).
- **Generator dengan `yield` (lazy)**: fungsi hanya memproses satu baris pada satu waktu. Ketika menemukan baris ERROR, fungsi menghasilkannya lalu **ditangguhkan** sambil menyimpan state lokalnya (posisi loop). Baris berikutnya baru diproses ketika diminta oleh `next()` atau `for`. Pada satu saat, memori hanya menyimpan satu baris yang sedang diproses, bukan seluruh hasil, sehingga pemakaian memori **hampir konstan** berapa pun jumlah barisnya (ukuran objek generator tetap kecil, seperti pada perbandingan 200 bytes vs 800.984 bytes di modul).
- Keuntungan tambahan: hasil pertama bisa langsung dipakai tanpa menunggu seluruh data selesai diproses, dan jika hanya butuh beberapa log pertama, sisa data tidak perlu diproses sama sekali.

*Catatan:* pada simulasi `server_logs` di atas, list-nya sendiri sudah ada di memori. Keuntungan nyata generator muncul ketika sumber datanya dibaca bertahap (misalnya baris demi baris dari file log), dan generator memastikan bahwa **hasil filter** tidak ikut dikumpulkan seluruhnya di memori.
